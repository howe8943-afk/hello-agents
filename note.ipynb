# Stage 0：理解 Agent
## 一、区分 chatbot、workflow、agent 和 multi-agent。

1. Chatbot（聊天机器人）
核心定位：基于自然语言的对话接口。

运行机制：被动响应用户的输入。它通常依赖预设的规则树（传统 NLP）或单一的大语言模型（LLM）来生成文本回复。

2. Workflow（工作流）
核心定位：确定性的任务编排系统。

运行机制：按照开发者预先定义好的节点（Nodes）和边（Edges）顺序执行。每个节点可能是一段代码、一个 API 调用，或者一次 LLM 的文本生成。


3. Agent（智能体）
核心定位：具备自主规划和执行能力的智能系统。

运行机制：以 LLM 为核心“大脑”，具备规划（Planning）、记忆（Memory）和工具调用（Tool Use/Action）三大能力。接收到一个宏大目标后，Agent 能自行拆解步骤，决定调用什么工具（如搜索网页、运行 Python 脚本、查询数据库），并根据工具返回的结果评估下一步行动，直到目标完成。


4. Multi-Agent（多智能体系统）
核心定位：多个独立 Agent 组成的分布式协作系统。

运行机制：将一个极度复杂的任务分配给不同“角色”的 Agent。例如，一个系统内包含“程序员 Agent”、“测试员 Agent”和“产品经理 Agent”。它们通过预设的通信机制（如顺序接力、群聊广播、层级汇报）进行信息交换、相互验证和辩论。

## 二、transformer的基础知识

这份文本主要解析了 Transformer 架构的核心模块，以及现代大语言模型广泛采用的 Decoder-Only 架构的工作原理。以下是核心内容的提炼总结：

### 1. Transformer 架构基础与核心机制

Transformer 彻底放弃了传统的循环网络（RNN/LSTM），依靠注意力机制实现了大规模并行计算，解决了长序列处理的效率瓶颈。最初的 Transformer 采用 **Encoder-Decoder（编码器-解码器）** 结构：编码器负责提取完整输入的上下文特征，解码器负责根据前文和编码器输出生成目标序列。

其核心模块包含：

* **自注意力机制 (Self-Attention)**：通过引入查询 (Query)、键 (Key)、值 (Value) 三个向量，计算词与词之间的相关性得分，使模型在处理每个词时能动态融合整个句子的上下文信息。
* **多头注意力 (Multi-Head Attention)**：将注意力计算在维度上切分为多个独立的“头”并行处理，使模型能够同时捕捉多种不同的特征关系（如时态、指代等），最后再进行整合。
* **前馈神经网络 (FFN)**：作用于每个位置的词元，通过先升维再降维的线性变换结合 ReLU 激活函数，进一步提取高阶特征。
* **残差连接与层归一化 (Add & Norm)**：残差连接用于防止网络加深时的梯度消失；层归一化用于稳定各层输入的数据分布，加速模型收敛并提高训练稳定性。
* **位置编码 (Positional Encoding)**：由于注意力机制本身无法感知词的顺序，通过固定的正余弦公式为每个词元向量叠加位置信息，使模型能够区分具有相同含义但在不同位置的词。

### 2. Decoder-Only 架构

Decoder-Only 架构是对 Transformer 的简化，它完全抛弃了编码器，是现代主流大语言模型（如 GPT、Llama）的基础架构。

* **自回归工作模式 (Autoregressive)**：模型的核心任务被简化为基于已知的前文序列，预测下一个最有可能出现的词，并将生成的词加入输入中继续预测。
* **掩码自注意力 (Masked Self-Attention)**：在训练时，通过因果掩码（将未来位置的注意力得分设为极小值）屏蔽当前词之后的信息。这从数学上阻止了模型“偷看”后续内容，确保了训练与生成阶段逻辑的严格一致。
* **架构优势**：训练目标统一且极其单纯（即预测下一个词），这使得模型更容易在海量无标注数据上扩展规模（Scaling），并且天然契合各类文本生成任务。
